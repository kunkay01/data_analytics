# Exploratory data analysis (EDA) of supermarket data

## Libraries and settings

```python
# Libraries
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
import pylab as py

# seaborn graphics settings
sns.set(color_codes=True)

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")

# Show current working directory
print(os.getcwd())
```

## Univariate non-graphical exploratory data analysis (EDA)

### Importing the enriched apartment data

```python
# Read and select variables
df_orig = pd.read_csv("supermarkets_data_enriched.csv")[['type',
                                                        'id',
                                                        'lat',
                                                        'lon',
                                                        'brand',
                                                        'shop',
                                                        'addr:city', 
                                                        'addr:housenumber', 
                                                        'addr:postcode', 
                                                        'geometry', 
                                                        'index_right',
                                                        'bfs_name',
                                                        'bfs_number',
                                                        'pop',
                                                        'pop_dens',
                                                        'frg_pct',
                                                        'emp']]

# Remove duplicates
df_orig = df_orig.drop_duplicates()
df_orig.head(5)

# Remove missing values
df_orig = df_orig.dropna()
df_orig.head(5)
```

### Count Number of supermarkets per brand

```python
brand_counts = df_orig['brand'].value_counts()
print(brand_counts)
```

### barchart with the number of supermarkets per brand

```python
plt.figure(figsize=(10, 12))
brand_counts.sort_values().plot(kind='barh')
plt.xlabel("Number of supermarkets")
plt.ylabel("Brand")
plt.title("Number of supermarkets per brand")
plt.tight_layout()
plt.show()
```

### scatterplotmatrix of the numeric variables lat, lon, pop, pop_dens, frg_pct, emp

```python
vars_num = ['lat', 'lon', 'pop', 'pop_dens', 'frg_pct', 'emp']

g = sns.PairGrid(df_orig[vars_num])
g.map_diag(plt.hist)
g.map_offdiag(plt.scatter)

g.map_upper(sns.scatterplot,color='darkgreen', s=10)
g.map_lower(sns.scatterplot, color='darkgreen', s=10)
g.map_diag(plt.hist, color='brown')

plt.show()
```

### Plot with the locations of supermarkets in diffrent colors

```python
brands_to_plot = ['Denner', 'Volg', 'Landi']
df_sub = df_orig.loc[df_orig['brand'].isin(brands_to_plot)]

# color mapping
colors = {'Denner': 'red',
          'Volg': 'blue',
          'Landi': 'green'}

plt.scatter(df_sub['lon'],
            df_sub['lat'],
            c=df_sub['brand'].map(colors))

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Locations of supermarkets by brand")
plt.tight_layout()
plt.show()
```
